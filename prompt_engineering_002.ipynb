```python
from google.colab import userdata
import requests
```

```python
OPENROUTER_API_KEY = userdata.get('API_KEY')


OPENROUTER_API_BASE = "https://openrouter.ai/api/v1"
MODELS = [
    "google/gemma-4-26b-a4b-it:free"
]


print(" Started Prompt Engineering Project!!!!.")
if OPENROUTER_API_KEY:
   print(f"Key preview: {OPENROUTER_API_KEY[:8]}{'*' * 24}")
else:
   print("OPENROUTER_API_KEY not found. Add it under Colab Secrets.")
```

```python
def call_llm(model, messages, temperature=0.3, max_tokens=300, system=None):

    full_messages = []

    if system:
        full_messages.append({
            "role": "system",
            "content": system
        })

    full_messages.extend(messages)

    headers = {
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "model": model,
        "messages": full_messages,
        "temperature": temperature,
        "max_tokens": max_tokens
    }

    response = requests.post(
        f"{OPENROUTER_API_BASE}/chat/completions",
        headers=headers,
        json=payload
    )

    return response.json()
```

```python
SYSTEM_PROMPT = """You are an AI academic content writer for an online learning platform.

Task:
Generate a concise course module summary using the provided input data.

Instructions:
1. Read the module name, course name, target audience, and key topics.
2. Write one concise summary paragraph explaining what the module covers.
3. Mention all listed key topics naturally in the summary.
4. Ensure the language is informative, neutral, and easy to understand.

Hard constraints:
- Maximum 80 words.
- Neutral and informative tone only.
- Must include all listed key topics.
- No marketing language or exaggerated claims.

Output format:
Return one paragraph only, under 80 words."""

user_prompt = """Module name: Introduction to Neural Networks
Course name: Machine Learning Fundamentals
Target audience: Working professionals with no prior ML experience
Key topics covered: perceptrons, activation functions, forward propagation, loss functions"""

result = call_llm(
    model=MODELS[0],
    system=SYSTEM_PROMPT,
    messages=[
        {
            "role": "user",
            "content": user_prompt
        }
    ],
    temperature=0.2,
    max_tokens=300
)

output = result["choices"][0]["message"]["content"]
print(output)

word_count = len(output.split())
print("Word count:", word_count)

if word_count <= 80:
    print("PASS ✅")
else:
    print("FAIL ❌")
```

```python
SYSTEM_PROMPT = """You are an AI customer support representative for an online learning platform.

Task:
Generate a customer support response using the provided issue details.

Instructions:
1. Begin with a polite greeting using the customer’s name.
2. Acknowledge the issue with empathy.
3. Clearly explain the resolution and expected timeline.
4. Close positively and invite further questions.

Hard constraints:
- Must follow this exact order: greet → acknowledge with empathy → explain resolution → positive close.
- Do not mention internal systems, teams, or escalation procedures.
- Tone must be professional, empathetic, and reassuring.

Output format:
Return exactly four short paragraphs or four clearly separated sentences."""

user_prompt = """Customer name: Customer B
Issue reported: unable to access the course materials after payment
Resolution: access will be restored within 2 hours; customer will receive a confirmation email"""

result = call_llm(
    model=MODELS[0],
    system=SYSTEM_PROMPT,
    messages=[
        {
            "role": "user",
            "content": user_prompt
        }
    ],
    temperature=0.2,
    max_tokens=300
)

output = result["choices"][0]["message"]["content"]
print(output)

checks = {
    "Greeting": "hello" in output.lower(),
    "Empathy": "sorry" in output.lower(),
    "Resolution": "2 hours" in output,
    "Positive close": "thank" in output.lower()
}

print(checks)

if all(checks.values()):
    print("PASS ✅")
else:
    print("FAIL ❌")
```
